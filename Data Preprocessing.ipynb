```python
# import libraries 
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
```

```python
# load in data
ps_raw = pd.read_csv(r"PS_2026.04.03_18.36.56.csv", skiprows = 292)
display(ps_raw)

# define only variables of interest
ps = ps_raw[
    [
        "pl_name", # planet name 
        "hostname", # star name 
        "default_flag", # recommended rows for duplicates
        "pl_rade", "pl_radeerr1", "pl_radeerr2", # planet radius 
        "pl_orbper", "pl_orbpererr1", "pl_orbpererr2", # orbital period 
        "pl_eqt", "pl_eqterr1", "pl_eqterr2", # equilibrium temperature 
        "st_teff", "st_tefferr1", "st_tefferr2", # stellar temperature 
        "st_rad", "st_raderr1", "st_raderr2", # stellar radius 
        "st_mass", "st_masserr1", "st_masserr2" # stellar mass
    ]
].copy()
display(ps)
```

```python
# convert to numeric
numeric_cols = [
        "pl_rade", "pl_radeerr1", "pl_radeerr2", 
        "pl_orbper", "pl_orbpererr1", "pl_orbpererr2", 
        "pl_eqt", "pl_eqterr1", "pl_eqterr2", 
        "st_teff", "st_tefferr1", "st_tefferr2", 
        "st_rad", "st_raderr1", "st_raderr2", 
        "st_mass", "st_masserr1", "st_masserr2"
]
for col in numeric_cols:
    ps[col] = pd.to_numeric(ps[col], errors = "coerce")
print(ps.dtypes)

# determine missing values
required_cols = [
    "pl_rade",
    "pl_orbper",
    "pl_eqt",
    "st_teff",
    "st_rad",
    "st_mass"
]
print("\nMissing Values:")
print(ps[required_cols].isnull().sum())
print("\nPercent Missing:")
print((ps[required_cols].isnull().sum() / len(ps) * 100).round(2))

# drop missing values in required fields
rows_before = ps.shape[0]
ps = ps.dropna(subset = required_cols).copy()
rows_after = ps.shape[0]
print("\nRows Before Dropping Missing Values:", rows_before)
print("Rows After Dropping Missing Values:", rows_after)
print("Rows Removed:", rows_before - rows_after)

# keep only default parameter rows
print("\nDefault Flag Distribution:")
print(ps["default_flag"].value_counts(dropna=False))
ps = ps[ps["default_flag"] == 1].copy()
```

```python
# create average uncertainty magnitude from asymmetric bounds
ps["pl_rade_unc"] = (ps["pl_radeerr1"].abs() + ps["pl_radeerr2"].abs()) / 2
ps["pl_orbper_unc"] = (ps["pl_orbpererr1"].abs() + ps["pl_orbpererr2"].abs()) / 2
ps["pl_eqt_unc"] = (ps["pl_eqterr1"].abs() + ps["pl_eqterr2"].abs()) / 2
ps["st_teff_unc"] = (ps["st_tefferr1"].abs() + ps["st_tefferr2"].abs()) / 2
ps["st_rad_unc"] = (ps["st_raderr1"].abs() + ps["st_raderr2"].abs()) / 2
ps["st_mass_unc"] = (ps["st_masserr1"].abs() + ps["st_masserr2"].abs()) / 2

# if uncertainty missing, fill with 0
ps["pl_rade_unc"] = ps["pl_rade_unc"].fillna(0)
ps["pl_orbper_unc"] = ps["pl_orbper_unc"].fillna(0)
ps["pl_eqt_unc"] = ps["pl_eqt_unc"].fillna(0)
ps["st_teff_unc"] = ps["st_rad_unc"].fillna(0)
ps["st_rad_unc"] = ps["st_rad_unc"].fillna(0)
ps["st_mass_unc"] = ps["st_mass_unc"].fillna(0)
```

```python
# create proxy habitability label
ps["proxy_habitable"] = np.where(
    (ps["pl_rade"] <= 1.6) &
    (ps["pl_eqt"] >= 180) &
    (ps["pl_eqt"] <= 310),
    1,
    0
)
display(ps[["pl_name", "pl_rade", "pl_eqt", "proxy_habitable"]])
```

```python
# inspect class balance
print(ps["proxy_habitable"].value_counts())
print("\nClass proportions:")
print(ps["proxy_habitable"].value_counts(normalize=True).round(4))

# summary stats
ps[
    [
        "pl_rade", "pl_orbper", "pl_eqt",
        "st_teff", "st_rad", "st_mass",
        "pl_rade_unc", "pl_orbper_unc", "pl_eqt_unc",
        "st_teff_unc", "st_rad_unc", "st_mass_unc",
        "proxy_habitable"
    ]
].describe()
```

```python
# export cleaned data
ps.to_csv("exoplanet_volatility_cleaned.csv", index = False)
```
